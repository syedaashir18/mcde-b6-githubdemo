```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]


averages = {
    students[0]["name"]: sum(students[0]["marks"]) / len(students[0]["marks"]),
    students[1]["name"]: sum(students[1]["marks"]) / len(students[1]["marks"]),
    students[2]["name"]: sum(students[2]["marks"]) / len(students[2]["marks"])
}


top_student = max(averages, key=averages.get)


names = (students[0]["name"], students[1]["name"], students[2]["name"])


unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])

print("Averages:", averages)
print("Top student:", top_student)
print("Names:", names)
print("Unique marks:", unique_marks)
```

```python
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}


common = employees["Ali"] & employees["Sara"] & employees["Ahmed"]


ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]


all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]


required = {"Python", "SQL"}
meets_requirement = {
    "Ali": required.issubset(employees["Ali"]),
    "Sara": required.issubset(employees["Sara"]),
    "Ahmed": required.issubset(employees["Ahmed"])
}

print("Common:", common)
print("Ali only:", ali_only)
print("All skills:", all_skills)
print("Meets Python+SQL:", meets_requirement)
```

```python
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

(n1, p1, q1), (n2, p2, q2), (n3, p3, q3) = sales

revenue_dict = {n1: p1 * q1, n2: p2 * q2, n3: p3 * q3}

total_revenue = sum(revenue_dict.values())
top_product = max(revenue_dict, key=revenue_dict.get)
unique_prices = {p1, p2, p3}

print("Revenue dict:", revenue_dict)
print("Total revenue:", total_revenue)
print("Top product:", top_product)
print("Unique prices:", unique_prices)
```

```python
customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

spending = {
    customers["C101"]["name"]: sum(customers["C101"]["orders"]),
    customers["C102"]["name"]: sum(customers["C102"]["orders"]),
    customers["C103"]["name"]: sum(customers["C103"]["orders"])
}

top_customer = max(spending, key=spending.get)

totals_tuple = (
    sum(customers["C101"]["orders"]),
    sum(customers["C102"]["orders"]),
    sum(customers["C103"]["orders"])
)

print("Spending:", spending)
print("Top customer:", top_customer)
print("Totals tuple:", totals_tuple)
```

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

py, sql, aws = courses["Python"], courses["SQL"], courses["AWS"]

all_three = py & sql & aws
at_least_two = (py & sql) | (py & aws) | (sql & aws)
exactly_two = at_least_two - all_three
exactly_one = (py - sql - aws) | (sql - py - aws) | (aws - py - sql)
unique_count = len(py | sql | aws)

print("Exactly two:", exactly_two)
print("All three:", all_three)
print("Exactly one:", exactly_one)
print("Unique students:", unique_count)
```

```python
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]


(a1, ap1, aq1), (a2, ap2, aq2) = transactions[0]["items"]
(b1, bp1, bq1), (b2, bp2, bq2) = transactions[1]["items"]
(c1, cp1, cq1) = transactions[2]["items"][0]

totals = {
    101: ap1 * aq1 + ap2 * aq2,
    102: bp1 * bq1 + bp2 * bq2,
    103: cp1 * cq1
}

total_revenue = sum(totals.values())
ali_spending = totals[101] + totals[103]
largest_txn = max(totals, key=totals.get)
unique_products = {a1, a2, b1, b2, c1}

print("Total revenue:", total_revenue)
print("Ali spending:", ali_spending)
print("Largest transaction ID:", largest_txn)
print("Unique products:", unique_products)
```

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": [1000, 1200, 1100]},
    "pipeline_2": {"status": "failed",  "records": [500, 0, 0]},
    "pipeline_3": {"status": "success", "records": [1700, 1600, 1500]}
}

totals = {
    "pipeline_1": sum(pipelines["pipeline_1"]["records"]),
    "pipeline_2": sum(pipelines["pipeline_2"]["records"]),
    "pipeline_3": sum(pipelines["pipeline_3"]["records"])
}

total_processed = sum(totals.values())
largest_pipeline = max(totals, key=totals.get)

averages = {
    "pipeline_1": totals["pipeline_1"] / len(pipelines["pipeline_1"]["records"]),
    "pipeline_2": totals["pipeline_2"] / len(pipelines["pipeline_2"]["records"]),
    "pipeline_3": totals["pipeline_3"] / len(pipelines["pipeline_3"]["records"])
}


failed = []
if pipelines["pipeline_1"]["status"] == "failed":
    failed.append("pipeline_1")
if pipelines["pipeline_2"]["status"] == "failed":
    failed.append("pipeline_2")
if pipelines["pipeline_3"]["status"] == "failed":
    failed.append("pipeline_3")

print("Total processed:", total_processed)
print("Failed:", failed)
print("Largest pipeline:", largest_pipeline)
print("Averages:", averages)
```

```python
from bisect import bisect_left

accounts = {
    "A101": {"holder": "Ali",  "transactions": [5000, -2000, 1500]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, 2000]},
    "A103": {"holder": "Omar", "transactions": [4000, -1500, 2500]}
}


balances = {
    "A101": sum(accounts["A101"]["transactions"]),
    "A102": sum(accounts["A102"]["transactions"]),
    "A103": sum(accounts["A103"]["transactions"])
}
highest_account = max(balances, key=balances.get)


all_tx = (accounts["A101"]["transactions"]
        + accounts["A102"]["transactions"]
        + accounts["A103"]["transactions"])
sorted_tx = sorted(all_tx)


cut = bisect_left(sorted_tx, 0)
total_withdrawals = abs(sum(sorted_tx[:cut]))
largest_deposit = max(all_tx)

print("Final balances:", balances)
print("Highest balance:", highest_account, balances[highest_account])
print("Total withdrawals:", total_withdrawals)
print("Largest deposit:", largest_deposit)
```

```python
users = {
    "Ali":   {"purchased": {"Laptop", "Mouse"},    "wishlist": {"Keyboard", "Monitor"}},
    "Sara":  {"purchased": {"Keyboard", "Mouse"},  "wishlist": {"Laptop", "Monitor"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

p1, p2, p3 = users["Ali"]["purchased"], users["Sara"]["purchased"], users["Ahmed"]["purchased"]
w1, w2, w3 = users["Ali"]["wishlist"], users["Sara"]["wishlist"], users["Ahmed"]["wishlist"]


bought_by_two = (p1 & p2) | (p1 & p3) | (p2 & p3)


wishlist_not_bought = (w1 | w2 | w3) - (p1 | p2 | p3)


shared_wishlist = (w1 & w2) | (w1 & w3) | (w2 & w3)


all_products = p1 | p2 | p3 | w1 | w2 | w3

print("Bought by 2+:", bought_by_two)
print("Wishlist not bought:", wishlist_not_bought)
print("Shared wishlist:", shared_wishlist)
print("All products:", all_products)
```

```python
company = {
    "departments": {
        "Engineering": {
            "employees": {
                "Ali":  {"salary": 120000, "skills": {"Python", "AWS", "SQL"}},
                "Sara": {"salary": 100000, "skills": {"Python", "Databricks", "SQL"}}
            },
            "budget": 300000
        },
        "Analytics": {
            "employees": {
                "Ahmed": {"salary": 90000, "skills": {"SQL", "Snowflake", "Python"}},
                "Zain":  {"salary": 80000, "skills": {"Python", "SQL", "Tableau"}}
            },
            "budget": 200000
        }
    }
}

eng = company["departments"]["Engineering"]
ana = company["departments"]["Analytics"]

eng_payroll = eng["employees"]["Ali"]["salary"] + eng["employees"]["Sara"]["salary"]
ana_payroll = ana["employees"]["Ahmed"]["salary"] + ana["employees"]["Zain"]["salary"]

eng_remaining = eng["budget"] - eng_payroll
ana_remaining = ana["budget"] - ana_payroll

eng_skills = eng["employees"]["Ali"]["skills"] | eng["employees"]["Sara"]["skills"]
ana_skills = ana["employees"]["Ahmed"]["skills"] | ana["employees"]["Zain"]["skills"]
common_skills = eng_skills & ana_skills

print("Payroll:", eng_payroll, ana_payroll)
print("Remaining budget:", eng_remaining, ana_remaining)
print("Common skills:", common_skills)
```
