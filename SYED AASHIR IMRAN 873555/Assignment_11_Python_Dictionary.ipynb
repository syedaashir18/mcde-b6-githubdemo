```python
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)
top_student = max(marks, key=marks.get)

print("Highest:", highest)
print("Lowest:", lowest)
print("Average:", average)
print("Top student:", top_student)
```

```python
employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}


print("Sara salary:", employees["E102"]["salary"])


employees["E101"]["salary"] = employees["E101"]["salary"] * 1.15
print("Ali new salary:", employees["E101"]["salary"])

# Total payroll
total_payroll = (employees["E101"]["salary"]
               + employees["E102"]["salary"]
               + employees["E103"]["salary"])
print("Total payroll:", total_payroll)
```

```python
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}


laptop_sub = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse_sub = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard_sub = order["keyboard"]["price"] * order["keyboard"]["quantity"]


grand_total = laptop_sub + mouse_sub + keyboard_sub
after_discount = grand_total - (grand_total * 0.10)

print("Laptop subtotal:", laptop_sub)
print("Mouse subtotal:", mouse_sub)
print("Keyboard subtotal:", keyboard_sub)
print("Grand total:", grand_total)
print("After 10% discount:", after_discount)
```

```python
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}


averages = {
    "Ali": sum(employees["Ali"].values()) / len(employees["Ali"]),
    "Sara": sum(employees["Sara"].values()) / len(employees["Sara"]),
    "Ahmed": sum(employees["Ahmed"].values()) / len(employees["Ahmed"])
}
print("Averages:", averages)


top_performer = max(averages, key=averages.get)
print("Top performer:", top_performer)   # Sara


sql_scores = {
    "Ali": employees["Ali"]["sql"],
    "Sara": employees["Sara"]["sql"],
    "Ahmed": employees["Ahmed"]["sql"]
}
best_sql_name = max(sql_scores, key=sql_scores.get)
print("Highest SQL:", sql_scores[best_sql_name], "by", best_sql_name)
```

```python
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

values = {
    "P101": inventory["P101"]["price"] * inventory["P101"]["stock"],
    "P102": inventory["P102"]["price"] * inventory["P102"]["stock"],
    "P103": inventory["P103"]["price"] * inventory["P103"]["stock"]
}

total_value = sum(values.values())
print("Total value:", total_value)

top_id = max(values, key=values.get)
print("Most valuable:", inventory[top_id]["name"])

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print("Laptop stock:", inventory["P101"]["stock"])
print("Mouse stock:", inventory["P102"]["stock"])
```
