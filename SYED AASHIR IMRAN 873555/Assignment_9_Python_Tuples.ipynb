```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 90))
)

sara_second = students[1][1][1]

ali_scores = students[0][1]
ali_avg = sum(ali_scores) / len(ali_scores)

ahmed_highest = max(students[2][1])

print("Sara second score:", sara_second)
print("Ali average:", round(ali_avg, 2))
print("Ahmed highest:", ahmed_highest)
```

```python
from operator import itemgetter

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = tuple(sorted(employees, key=itemgetter(1), reverse=True))

print(sorted_employees)
```

```python
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])


product = diagonal[0] * diagonal[1] * diagonal[2]


row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))

print("Diagonal:", diagonal)
print("Product:", product)
print("Row sums:", row_sums)
```

```python
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

names, prices, quantities = zip(*records)

total_value = (prices[0] * quantities[0]
             + prices[1] * quantities[1]
             + prices[2] * quantities[2])

expensive = names[prices.index(max(prices))]


reversed_names = names[::-1]

print("Inventory value:", total_value)
print("Most expensive:", expensive)
print("Reversed names:", reversed_names)
```
