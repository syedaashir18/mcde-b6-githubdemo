```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union:", A | B)
print("Intersection:", A & B)
print("A - B:", A - B)
print("Symmetric difference:", A ^ B)
```

```python
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hina", "Usman"}

both = python_students & sql_students
exactly_one = python_students ^ sql_students

print("Both courses:", both)
print("Exactly one course:", exactly_one)
```

```python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Spark"}

print("Subset:", required_skills.issubset(candidate_skills))
print("Superset:", candidate_skills.issuperset(required_skills))
```

```python
numbers = {1, 2, 4, 7, 9, 10}

full_set = set(range(1, 11))
missing = full_set - numbers

print("Missing:", missing)
```

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hina"}
aws = {"Ahmed", "Sara", "Usman"}


all_three = python & sql & aws


only_py_sql = (python & sql) - aws


at_least_two = (python & sql) | (python & aws) | (sql & aws)

print("All three:", all_three)
print("Python and SQL, not AWS:", only_py_sql)
print("At least two:", at_least_two)
```

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}


all_days = day1 & day2 & day3


exactly_one = (day1 - day2 - day3) | (day2 - day1 - day3) | (day3 - day1 - day2)


at_least_two = (day1 & day2) | (day1 & day3) | (day2 & day3)

print("All days:", all_days)
print("Exactly one day:", exactly_one)
print("At least two days:", at_least_two)
```

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions
extra = user_permissions - required_permissions
fully_authorized = required_permissions.issubset(user_permissions)

print("Missing:", missing)
print("Extra:", extra)
print("Fully authorized:", fully_authorized)
```

```python
website = {"C1", "C2", "C3", "C4"}
mobile = {"C2", "C3", "C5"}
store = {"C3", "C5", "C6"}

all_three = website & mobile & store

exactly_one = (website - mobile - store) | (mobile - website - store) | (store - website - mobile)

at_least_two = (website & mobile) | (website & store) | (mobile & store)
exactly_two = at_least_two - all_three

print("Exactly one:", exactly_one)
print("Exactly two:", exactly_two)
print("All three:", all_three)
```
