```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]


main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]

secondary_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]

middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]

main_sum = sum(main_diagonal)
secondary_sum = sum(secondary_diagonal)
middle_sum = sum(middle_column)

print("Main diagonal:", main_diagonal, "Sum =", main_sum)
print("Secondary diagonal:", secondary_diagonal, "Sum =", secondary_sum)
print("Middle column:", middle_column, "Sum =", middle_sum)
```

```python
data = [15, 5, 25, 10, 35, 20, 30]

unique_numbers = list(set(data))

unique_numbers.sort()

largest_three = unique_numbers[-3:]

smallest_two = unique_numbers[:2]

smallest_two.reverse()

result = largest_three + smallest_two

print("Result:", result)
```
